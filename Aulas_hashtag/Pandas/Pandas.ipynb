# Curso básico de Pandas

```python
import pandas as pd
```

## Criando um dataframe simples

```python
#dataframe = pd.DataFrame()
#Criando um dataframe a partir de dicionários do python
dic_vendas = {"Data da Venda": ['15/02/2021'],
              "valor": [500],
              "Produto": ["Feijão"],
              "Quantidade": [50]
}

tabela_vendas = pd.DataFrame(dic_vendas)
```

```python
print(tabela_vendas)
display(tabela_vendas)
```

## Criando dataframe a partir de base do excel

```python
tabela_vendas = pd.read_excel("Vendas.xlsx") #Cria o Dataframe
display(tabela_vendas)
```

```python
display(tabela_vendas.head(10))
display(tabela_vendas.describe())
```

Uma série do Pandas é quando se tem apenas uma única coluna. Dessa maneira, o python interpreta como sendo uma "Série do Pandas" e não uma tabela.
Quando tem mais de uma coluna, se torna uma tabela

```python
produtos = tabela_vendas["Produto"] #-> Exibe uma série do pandas
display(produtos)
```

```python
produtos_2 = tabela_vendas[["Produto", "Data"]]
display(produtos_2)
```

O método .loc() permite pegar os dados do Dataframe a partir de alguma condição ou valor específico

```python
tabela_vendas.loc[1:5] #Pegar uma linha específica 

#Pegar linhas que correspondem a uma condição. A comparação deve estar dentro dos colchetes do loc(linhas, colunas)
display(tabela_vendas.loc[tabela_vendas["ID Loja"] == "Norte Shopping", ["Data", "Produto", "Quantidade"]]) #Exibir apenas as linhas onde o ID da loja é igual à Norte Shopping
```

```python
#Pegar apenas as camisetas
display(tabela_vendas.loc[tabela_vendas["Produto"] == "Camiseta", ["Código Venda", "Quantidade", "Valor Final"]])
```
