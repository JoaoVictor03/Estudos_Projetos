## 1 Ler e visualizar dados

```python
#Ler e visualizar os dados
import pandas as pd

tabela_dataframe = pd.read_excel("dados.xlsx")
display(tabela_dataframe)
```

```python
# Exibir as 5 primeiras linhas
display(tabela_dataframe.head(5))

#Exibir o total de linhas e colunas usando o método Len para medir o "comprimento da lista"
total_linhas = len(tabela_dataframe.index) #-> Usando Index
total_colunas = len(tabela_dataframe.columns)

display(total_linhas, total_colunas)


#Exibir 
```

```python
#Exibir o total de linhas e colunas usando o método shape
print(tabela_dataframe.shape)
```

## 2 Explorar o DataFrame

```python
# 2.1 Liste todos os nomes das colunas
#Usar os column.values para exibir os valores dos dados das colunas
nomes_colunas = tabela_dataframe.columns.values 
print(nomes_colunas)

#Usar o método columns.tolist()
nomes_colunas = tabela_dataframe.columns.to_list()
print(nomes_colunas)
```

```python
# 2.2 Exiba as informações gerais (tipos de dados, nulos, etc.)
# Para isso, usar o método info.()
informacoes = tabela_dataframe.info()
```

```python
# 2.3 Calcule a média de salários
media_salarios = tabela_dataframe["Salario"].mean()
print(media_salarios)
```

## 3 Filtragem de dados

```python
# 3.1 Mostre apenas os funcionários do departamento "Financeiro"
# Tabela, onde tabela[coluna] == "Parâmetro"
funcionarios_financeiro = tabela_dataframe[tabela_dataframe["Departamento"] == "Financeiro"]
display(funcionarios_financeiro)

# 3.2 Mostre quem ganha mais de R$4500
funcionarios_salarios = tabela_dataframe[tabela_dataframe["Salario"] > 4500]
display(funcionarios_salarios)


# 3.3 Mostre os funcionários com menos de 30 anos
funcionarios_anos = tabela_dataframe[tabela_dataframe["Idade"] < 30]
display(funcionarios_anos)
```

## 4 Ordenação de itens

```python
# 4.1 Ordene o DataFrame por salário (decrescente)
tabela_ordenada = tabela_dataframe.sort_values("Salario", ascending=False)
display(tabela_ordenada)

# 4.2 Ordene por idade (crescente)
tabela_ordenada_idades = tabela_dataframe.sort_values(by=["Idade"], ascending=True)
display(tabela_ordenada_idades)
```

# 5 Agrupamentos e estatísticas

```python
# 5.1 Calcule o salário médio por departamento
print("Exibir a média por departamento")
media_agrupada = tabela_dataframe.groupby("Departamento")["Salario"].mean()
display(media_agrupada)
```

```python
# 5.2 Mostre a idade média por departamento
idade_media = tabela_dataframe.groupby("Departamento")["Idade"].mean()
display(idade_media)
```
