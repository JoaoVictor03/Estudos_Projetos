```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
eixo_x = ["Janeiro", "Fevereiro", "Março", "Abril", "Maio"]
eixo_y = [120, 123, 142, 180, 162]
eixo_y2 = [120, 150, 120, 177, 162]

plt.plot(eixo_x, eixo_y, marker="s", color="Magenta", linestyle="dashed") #por padrão, um gráfico de linhas 2d é gerado
#padrão de formatação = marker|line|color
plt.plot()
```

```python
#Plotando um gráfico com mais de uma linha
meses = ["Janeiro", "Fevereiro", "Março", "Abril", "Maio"]
meses_2 = ["Janeiro", "Fevereiro", "Março", "Abril", "Maio"]
loja_1 = [100,120,115,199,220]
loja_2 = [111,110,134,170,200]

plt.plot(loja_1)
plt.plot(loja_2)
plt.show()

plt.plot(meses, loja_1, meses_2, loja_2)
plt.show()
```

```python
meses = ["Janeiro", "Fevereiro", "Março", "Abril", "Maio"]
loja_1 = [100,120,115,199,220]

plt.scatter(meses, loja_1, s=loja_1)
#Definir os labels e título de forma simples
plt.xlabel("Meses")
plt.ylabel("Vendas")
plt.title("Vendas para a loja 1 de acordo com os meses")
```
